```python
import json
import matplotlib.pyplot as plt
import numpy as np

# --- 1. SIMULIMI I LOGJIKËS SË ALDËS ---
# (Këtu do të kishim importet nga src/ranking/ dhe src/text/)
# Po e bëjmë një shembull të thjeshtë që mund ta ekzekutosh direkt:

def get_hybrid_scores(articles):
    # Shembull: Hybrid score si kombinim i relevancës dhe page_rank
    scores = []
    for art in articles:
        # Përafrohet me logjikën e skedarëve të tu .py
        score = art.get('relevance', 0) * 0.6 + art.get('pagerank', 0) * 0.4
        scores.append(score)
    return scores

# --- 2. NGARKIMI I TË DHËNAVE ---
# Në vend të këtij shembulli, ti mund të bësh: with open('articles.json') as f: data = json.load(f)
data = [
    {"title": "Artikulli A", "relevance": 0.8, "pagerank": 0.5},
    {"title": "Artikulli B", "relevance": 0.9, "pagerank": 0.2},
    {"title": "Artikulli C", "relevance": 0.4, "pagerank": 0.9},
    {"title": "Artikulli D", "relevance": 0.6, "pagerank": 0.7}
]

titles = [a['title'] for a in data]
scores = get_hybrid_scores(data)

# --- 3. VIZUALIZIMI (GRAFIKU PËR ALDËN) ---
# Ky është grafiku që e bën projektin të duket profesional
print("Duke gjeneruar grafikun e renditjes së artikujve...")

plt.figure(figsize=(10, 6))
bars = plt.bar(titles, scores, color='teal', alpha=0.7)
plt.xlabel("Titujt e Artikujve")
plt.ylabel("Hybrid Score (Renditja)")
plt.title("Analiza e Renditjes së Artikujve (Ranking)")
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Shtimi i vlerave mbi shtylla
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 0.01, round(yval, 2), ha='center', va='bottom')

plt.show()

# --- 4. ANALIZA STATISTIKORE ---
print(f"Artikulli më i rëndësishëm: {titles[np.argmax(scores)]}")
```
